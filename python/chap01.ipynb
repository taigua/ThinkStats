```python
preg = preg.with_columns(
    pl.when(pl.col("birthwgt_lb").is_in([51, 97, 98, 99]))
    .then(None)
    .otherwise(pl.col("birthwgt_lb"))
    .alias("birthwgt_lb")
)
```

```python
preg.get_column("agepreg").mean()
```

```python
preg = preg.with_columns(pl.col("agepreg") / 100.0)
preg.get_column("agepreg").mean()
```

```python
preg.get_column("birthwgt_oz").value_counts().sort("birthwgt_oz", nulls_last=True)
```

```python
preg = preg.with_columns(
    pl.when(pl.col("birthwgt_oz").is_in([97, 98, 99]))
    .then(None)
    .otherwise(pl.col("birthwgt_oz"))
    .alias("birthwgt_oz")
)
```

```python
preg = preg.with_columns(
    totalwgt_lb=pl.col("birthwgt_lb") + pl.col("birthwgt_oz") / 16.0
)
preg.get_column("totalwgt_lb").mean()
```

```python
weights = preg.get_column("totalwgt_lb")
n = weights.count()
n
```

```python
mean = weights.sum() / n
mean
```

```python
weights.mean()
```

```python
squared_deviations = (weights - mean) ** 2
```

```python
var = squared_deviations.sum() / n
var
```

```python
weights.var()
```

```python
weights.var(0)
```

```python
std = np.sqrt(var)
std
```

```python
weights.std(0)
```
